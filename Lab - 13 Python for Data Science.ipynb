### 1. Fetch MCA Faculty Page and Check Status
####  Send an HTTP GET request to https://darshan.ac.in/faculty-list/mca using requests.get().
#### Display the HTTP status code to verify a successful connection (200) and print the length of the page source code.

```python
import requests
res= requests.get('https://darshan.ac.in/faculty-list/mca')
print(res.status_code)
print(len(res.text))
```

### 2. Extract Web Page Title.
#### Parse the HTML content using BeautifulSoup with 'html.parser'.
#### Find and display the content of the title tag to identify the webpage.

```python
from bs4 import BeautifulSoup
soup = BeautifulSoup(res.text,"html.parser")
print(soup.title.text)
```

### 3. Scrape Faculty Names using Heading Tags.
#### Locate all faculty names displayed on the page (typically enclosed in h3 tag or specific heading tags within faculty cards).
#### Loop through the results and display the first 5 faculty names cleanly without leading or trailing spaces

```python
name  = soup.find_all('h2','g-mb-0')
names = []
for i in name:
    names.append(i.text.strip())
print(names)
```

### 4. Extract Faculty Designations.
#### Identify the HTML tag or class used for designations (e.g., Assistant Professor, Associate Professor, Head of Department).
#### Extract and print the designations of the first 5 faculty members.

```python
designation=soup.find_all('span','g-mb-10')
designations=[]
for i in designation:
    designations.append(i.text.strip())
print(designations)
```
