### 04) WAP to plot a line chart showing temperature variation during a week. Mark the lowest and highest temperatue.

```python
Days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
Temperature = [20, 32, 31, 33, 34, 43, 36]
```

```python
min_temp = min(Temperature)
max_temp = max(Temperature)
min_day = Days[Temperature.index(min_temp)]
max_day = Days[Temperature.index(max_temp)]

plt.plot(Days, Temperature, marker="o")
plt.scatter(min_day, min_temp, color="blue", s=100, label="Lowest")
plt.scatter(max_day, max_temp, color="red", s=100, label="Highest")
plt.text(min_day, min_temp - 2, str(min_temp))
plt.text(max_day, max_temp + 1, str(max_temp))
plt.xlabel("Days")
plt.ylabel("Temperature (°C)")
plt.title("Temperature variation during a week")
plt.legend()
plt.show()
```

### 05) WAP to create a detailed pie chart representing the monthly expenditure distribution of a family. 
### The chart should display labels, percentage values, title, different colors for each slice, and explode one slice to highlight the highest expense.

```python
Categories = ["Rent", "Food", "Education", "Transport", "Entertainment", "Savings"]
Amount = [15000, 8000, 6000, 3000, 2000, 5000]
```

```python
colors = ["red", "orange", "yellow", "green", "skyblue", "violet"]
explode = [0.1, 0, 0, 0, 0, 0]     # Rent is the highest expense

plt.pie(Amount, labels=Categories, colors=colors, explode=explode, autopct="%1.1f%%")
plt.title("Monthly Expenditure of a Family")
plt.show()
```

### 06) WAP to create bar chart showing the marks obtained by students in different subjects.

### 10) WAP to create a box plot representing the distribution of daily commuting time (in minutes) of employees and identify outliers (if any).

```python
CommutingTime = [5, 18, 20, 22, 25, 27, 28, 30, 32, 35, 38, 40, 42, 90]
```

```python
plt.boxplot(CommutingTime)
plt.ylabel("Commuting Time (minutes)")
plt.title("Daily Commuting Time of Employees")
plt.show()

# Outlier check using the 1.5 * IQR rule
data = sorted(CommutingTime)
n = len(data)
q1 = data[n // 4]
q3 = data[(3 * n) // 4]
iqr = q3 - q1
lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr
outliers = [v for v in data if v < lower or v > upper]
print("Outliers:", outliers)
```
