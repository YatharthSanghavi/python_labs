### 17) Email Validator

**Problem Definition**

Validate an email using string methods.
A valid email:
- contains exactly one '@'
- contains at least one '.'
- does not start or end with '@' or '.'
- contains no spaces.

**Sample**
```
Input: student@gmail.com
Output: Valid Email
```


### 18) Username Generator

**Problem Definition**

Generate username from full name and birth year
Generate a username by:
- removing extra spaces,
- converting the name to lowercase,
- replacing spaces with underscores,
- appending the last two digits of the birth year.

**Sample**

```
Input:
Name : Rahul Kumar Sharma
Birth Year : 2004

Output:
Username : rahul_kumar_sharma04
```

### 22) Product Code Analyzer

**Problem Definition**

A product code is in the format ABC-1234-XY. Write a program to:
- Separate all three parts.
- Verify that:
        - first part contains only alphabets,
        - second part contains only digits,
        - third part contains only uppercase letters.

**Sample**

```
Input:
ABC-1234-XY

Output:
Code : ABC
Number : 1234
Category : XY
Valid Product Code
```
